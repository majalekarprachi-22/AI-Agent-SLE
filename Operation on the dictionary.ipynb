```python
# Nested Dictionary
```

```python
myfamily = {
  "child1" : {
    "name" : "Anil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Sunil",
    "year" : 2007
  },
  "child3" : {
    "name" : "Ram",
    "year" : 2011
  }
}

print(myfamily)
```

```python
child1 = {
  "name" : "Anil",
  "year" : 2004
}
child2 = {
  "name" : "Sunil",
  "year" : 2007
}
child3 = {
  "name" : "Ram",
  "year" : 2011
}

myfamily = {
  "child1" : child1,
  "child2" : child2,
  "child3" : child3
}

print(myfamily)
```

```python
myfamily = {
  "child1" : {
    "name" : "Anil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Sunil",
    "year" : 2007
  },
  "child3" : {
    "name" : "Ram",
    "year" : 2011
  }
}

print(myfamily["child2"]["name"])
```
