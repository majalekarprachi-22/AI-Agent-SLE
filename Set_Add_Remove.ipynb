```python
# Add and Remove Elements from a Set

numbers = {10, 20, 30, 40, 50}

# Display original set
print("Original Set:", numbers)

# Add an element
numbers.add(60)
print("After Adding:", numbers)

# Remove an element
numbers.remove(20)
print("After Removing:", numbers)

# Discard an element
numbers.discard(100)
print("After Discarding:", numbers)

# Display final set
print("Final Set:", numbers)
```
