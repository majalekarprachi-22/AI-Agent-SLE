```python

keys = ["Name", "Age", "City", "Marks"]

values = ["Prachi", 20, "Kolhapur", 90]

student = dict(zip(keys, values))

print("Keys List:", keys)
print("Values List:", values)
print("Student Dictionary:", student)
```
